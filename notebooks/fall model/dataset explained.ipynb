# The 3 fall datasets, explained simply

Each dataset is a collection of recordings from small motion sensors. People wore the sensors and did either a **fall** (on a safe mattress) or a normal everyday activity (**ADL** = Activities of Daily Living). The sensors measure two things:
- **Accelerometer**: how fast the sensor is speeding up or slowing down (a fall impact shows up as a big spike).
- **Gyroscope**: how fast the sensor is rotating (a fall makes the body twist and tumble).

Our goal (SilverGuard) is a **wrist** sensor at **50 readings per second (50 Hz)**, so we compare every dataset to that.

## 1. WEDA-FALL: a smartwatch on the wrist

**How it was recorded**
- Volunteers wore a **Fitbit Sense smartwatch** on the wrist. It recorded at about **50 Hz**.
- 14 young people did **8 kinds of falls** (slip, trip, fainting while sitting...) and 11 activities. 11 older people (77-95 years old) did **daily activities only**, no falls.

**What the files look like**
- One small CSV file per recording and sensor, e.g. `F01/U01_R01_accel.csv` = fall type 1, user 1, repeat 1.
- Columns: time, x, y, z. A separate file lists when each fall starts and ends.

**Good to know**
- Closest match to our project (wrist, ~50 Hz).
- No older person falls in it.
- Accelerometer values are cut off at 4 g, so big impacts look flatter than they really were.

## 2. UMAFall: sensors on 5 places of the body

**How it was recorded**
- 19 volunteers (age 18-68) wore sensors on the **wrist, chest, waist, ankle**, and a **phone in the trouser pocket**.
- They did **3 kinds of falls** (forward, backward, sideways) and 12 daily activities (walking, jogging, hopping, lying down...).
- Each recording is about **15 seconds**.

**What the files look like**
- One CSV per recording. The top has a text header (person's age, which sensor is where), then rows of numbers: time, x, y, z, sensor type, sensor id.
- All sensors are mixed together in the same file.

**Good to know**
- We only need the **wrist** rows, but those are recorded at just **~20 readings per second**, which is slow for catching a quick impact.
- 20 files from one person were damaged (numbers broken by Excel). Use the **corrected zip** instead of the original ones.

## 3. FallAllD: the biggest one, 3 body positions

**How it was recorded**
- 15 volunteers wore small data-loggers on the **waist, wrist and neck** (the paper says which is which; the folder does not).
- **35 kinds of falls** and **44 daily activities**, 6,605 recordings in total.
- Every recording is exactly **20 seconds**, with the fall impact in the **middle (~10 s)**.

**What the files look like**
- 4 plain text files per recording, with no header and no time column:
  `A` accelerometer, `G` gyroscope, `M` magnetometer (compass), `B` barometer (air pressure).
- The name tells you everything: `S01_D1_A013_T01_A.dat` = subject 1, device 1, activity 13, trial 1, accelerometer.
- Values are **raw sensor counts**, not real units, so they must be converted (accelerometer: about 4096 counts = 1 g).
- Accel and gyro run at **238 readings per second**.

**Good to know**
- Cleanest and largest dataset, and fast enough to slow down to 50 Hz to copy our sensor.
- Because falls are always in the middle of the 20 s, a model could cheat by learning "big spike in the middle = fall". Cut the recordings into small windows to avoid this.
